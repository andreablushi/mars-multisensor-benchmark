## Subsurface ice benchmark

SWIM (Morgan, Putzig et al. 2025) scores from -1 to 1 how well the orbital data fit
buried ice between 60S and 60N, and every tile there is sampled on an 11 by 11 grid,
3 km apart, at 1 to 5 m and below 5 m.

```python
"""Read the ice labels."""

ice_labels = read_ice_labels()
```

### Where the tiles are icy

Every labelled tile at its centre, coloured by the depths where at least half its
grid reaches the threshold.

```python
"""Map every labelled tile by the depths it is icy at."""

tiles.ice_category_map_widget(ice_labels)
```
