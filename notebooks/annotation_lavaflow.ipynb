# Lava flow annotation

This notebook reviews every SHARAD crop where a buried lava flow reflector is looked for. Each crop is one track over one tile, drawn from its surface down beside its cluttergram. Each click saves its verdict to `data/<version>/lavaflow/verdicts_lavaflow.json` straight away, the version set in the dataset's `configs/version.yaml`.

## Setup

```python
"""Import the candidate crops, the verdicts so far, and what reviews them."""

from lavaflow.analysis.artifacts import read_lavaflow_verdicts
from lavaflow.labels import built_candidate_labels
from lavaflow.visualization import review

labels = built_candidate_labels()
verdicts = read_lavaflow_verdicts()
crops = {label.crop: label for label in labels}
```

## Reading a crop

Both panels share their axes: traces run along the track from left to right, delay rows run down.

- Left, the radargram, in dB of echo power. Only the 256 rows under the atmosphere are drawn. One row is 37.5 ns, about 1.9 m of rock at permittivity 9. Blank samples are the ones the noise gate removed.
- Right, the cluttergram: the echo that MOLA topography alone predicts, with no subsurface. Whatever it shows comes from the surface, off to the side of the track.
- The title gives the track, the tile, its latitude and "[from the paper]" when the paper names the track. The note above the panels gives the crop's verdict and the region's counts.

The bright line near the top is the surface echo. Judge everything under it.

## Verdicts

- Reflector: a line under the surface that the cluttergram does not show at the same traces and rows. It runs along many traces, lies flat or dips gently, and often rises into the surface where the flow thins out.
- None: nothing under the surface, or every line there has a twin in the cluttergram. A glow in the first rows under the surface that follows it everywhere is the echo's sidelobe, and is also none.
- Unsure: a line too faint or broken to follow, a cluttergram that matches only part of it, or a mostly blank crop.

Judge each crop alone. A track can show a reflector on one tile and none on the next, and a named track is not a reflector until you see one. A short line counts only if you can follow it. Only reflector and none enter the evaluation, so put any doubt under unsure.

## Stepping through

Pick a region, then use the arrows or click a verdict, which saves it and moves to the next crop. Clicking again on a reviewed crop replaces its verdict. Only the crops on disk are listed, so build or download the lava flow candidates first.

## Examples

All four are crops of track 5095_01 (`s_00509501`), which Meng et al. 2020 show in their Fig. 2b. They sort reflectors into three patterns: a single reflector, one that extends to the surface, and both together. The first two are below. The papers define no crop for none or unsure, so those two examples are my own reading of the cluttergram test.

### Reflector that extends to the surface

Traces 0 to 22: a line about 16 rows under the surface rises to the right and merges with it near trace 22, where the cluttergram is empty. This is the pattern of Meng's Fig. 2b.

```python
"""Draw a reflector that rises into the surface."""

review.crop_radargram_panel(crops["b223_c0188/s_00509501"])
```

### Single reflector

Traces 38 to 69: a line about 18 rows under the surface runs parallel to it, with nothing in the cluttergram there. The cluttergram's dark shape lower down, from row 2700, has no match on the left.

```python
"""Draw a single reflector parallel to the surface."""

review.crop_radargram_panel(crops["b224_c0187/s_00509501"])
```

### None

A glow under the surface fades with depth along the whole crop, and no line can be followed in it.

```python
"""Draw a crop with no reflector."""

review.crop_radargram_panel(crops["b225_c0186/s_00509501"])
```

### Unsure

Traces 15 to 27: a faint patch about 15 rows under the surface, with an empty cluttergram there. It is only about 12 traces long and too weak to follow as a line.
